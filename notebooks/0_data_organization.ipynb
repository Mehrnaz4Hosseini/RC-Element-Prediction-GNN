```python
def quick_organize():

    source_folder = "/Users/sarayetel/Desktop/Research/Graph/4_Implementation/dataset/testdata"
    target_folder =  "/Users/sarayetel/Desktop/Research/Graph/RC-Element-Prediction-GNN/data/test"

    # Create target
    os.makedirs(target_folder, exist_ok=True)
    
    # Get all files sorted alphabetically
    edge_files = sorted(Path(source_folder).glob("Edge-*.xlsx"))
    edge_files.extend(sorted(Path(source_folder).glob("Edge-*.xls")))
    
    feature_files = sorted(Path(source_folder).glob("Feature-*.xlsx"))
    feature_files.extend(sorted(Path(source_folder).glob("Feature-*.xls")))
    
    label_files = sorted(Path(source_folder).glob("Label-*.xlsx"))
    label_files.extend(sorted(Path(source_folder).glob("Label-*.xls")))
    
    # How many complete sets can we make?
    max_sets = min(len(edge_files), len(feature_files), len(label_files))
    
    print(f"Found: {len(edge_files)} Edge, {len(feature_files)} Feature, {len(label_files)} Label")
    print(f"Creating {max_sets} sample folders...\n")
    
    # Create folders
    for i in range(max_sets):
        sample_num = i + 1
        sample_name = f"sample_{sample_num}"
        sample_path = os.path.join(target_folder, sample_name)
        os.makedirs(sample_path, exist_ok=True)
        
        # Get files for this sample
        edge_file = edge_files[i]
        feature_file = feature_files[i]
        label_file = label_files[i]
        
        # Move files
        shutil.move(str(edge_file), os.path.join(sample_path, edge_file.name))
        shutil.move(str(feature_file), os.path.join(sample_path, feature_file.name))
        shutil.move(str(label_file), os.path.join(sample_path, label_file.name))
        
        print(f"✅ sample_{sample_num}:")
        print(f"   {edge_file.name}")
        print(f"   {feature_file.name}")
        print(f"   {label_file.name}\n")
    
    print(f"Done! Created {max_sets} sample folders")
    
    # Count remaining
    remaining = (len(edge_files) - max_sets) + (len(feature_files) - max_sets) + (len(label_files) - max_sets)
    if remaining > 0:
        print(f"\n⚠️  {remaining} files remain unorganized (incomplete sets)")

quick_organize()
```
